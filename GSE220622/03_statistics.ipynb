# Load packages

```python
import pandas as pd
from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests
from scipy.stats import mannwhitneyu
import statsmodels.formula.api as smf
```

# Load data

```python
path = "path/to/GSE220622"

df_data = pd.read_excel(f"{path}/data.xlsx", index_col=0)
df_feats = pd.read_excel(f"{path}/feats.xlsx", index_col=0)

age_accs = df_feats.index[df_feats['Type'] == 'Epigenetic Age Acceleration'].to_list()
metrics = df_feats.index[df_feats['Type'] == 'Epigenetic Metric'].to_list()
cells = df_feats.index[df_feats['Type'] == 'Cell'].to_list()
episcores = df_feats.index[df_feats['Type'] == 'EpiScore'].to_list()
```

# Statistical analysis

## Epigenetic Age Acceleration

```python
df_stat = pd.DataFrame(index=age_accs)
    
for age_acc in age_accs:
    # Mann-Whitney U test
    df_stat.at[age_acc, "mw_stat"], df_stat.at[age_acc, "mw_pval"] = mannwhitneyu(df_data.loc[df_data['Status'] == 'Control', age_acc], df_data.loc[df_data['Status'] == 'Case', age_acc], alternative='two-sided')
    
    # ANCOVA with age as covariate
    ancova_age = smf.ols(formula=f"Q('{age_acc}') ~ Q('Status') + Age", data=df_data).fit()
    feat_trgt = ancova_age.params.index[ancova_age.params.index.str.contains('Status')].values[0]
    df_stat.at[age_acc, "ancova_age_tval"] = ancova_age.tvalues[feat_trgt]
    df_stat.at[age_acc, "ancova_age_pval"] = ancova_age.pvalues[feat_trgt]
    
# Multiple testing correction
_, df_stat.loc[age_accs, "mw_pval_fdr_bh"], _, _ = multipletests(df_stat.loc[age_accs, "mw_pval"].values, 0.05, method='fdr_bh')
_, df_stat.loc[age_accs, "ancova_age_pval_fdr_bh"], _, _ = multipletests(df_stat.loc[age_accs, "ancova_age_pval"].values, 0.05, method='fdr_bh')

# Save results
df_stat.sort_values(['mw_pval'], ascending=[True], inplace=True)
df_stat.index = df_stat.index.str.replace(' acceleration', '')
df_stat.to_excel(f"{path}/stat_age_accs.xlsx")
```
